```python
# Importamos librerías necesarias

import pandas as pd          # Para manejar datos (tablas)
import numpy as np           # Operaciones numéricas
from scipy import stats      # Pruebas estadísticas (t de Student)
import matplotlib.pyplot as plt  # Gráficos
```

```python
import pandas as pd

df = pd.read_csv("../dataset_general/datos.csv")

df.head()
```

```python
# Separar eficiencia

manual = df[df["metodo"] == "manual"]["eficiencia"]
ia = df[df["metodo"] == "ia"]["eficiencia"]

print("Cantidad manual:", len(manual))
print("Cantidad IA:", len(ia))
```

```python
import numpy as np

media_manual = np.mean(manual)
media_ia = np.mean(ia)

print("=== EFICIENCIA ===")
print(f"Manual: {media_manual:.2f}")
print(f"IA: {media_ia:.2f}")
```

```python
from scipy import stats

t_stat, p_value = stats.ttest_ind(ia, manual)

print("t =", t_stat)
print("p =", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Se RECHAZA H0 → La IA mejora la eficiencia")
else:
    print("NO se rechaza H0")
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.boxplot([manual, ia], tick_labels=["Manual", "IA"])

plt.title("Eficiencia del proceso")
plt.ylabel("Eficiencia")

plt.grid()
plt.show()
```
